## 4. Input features (Review 1 — Step 1.3)

**Primary text input**
- `full_text` — the student's essay. This is the main input to the model.

**Potential contextual/auxiliary inputs**
- `assignment` — the writing prompt/task instructions the student was responding to.
- `prompt_name` — a short identifier/name for the prompt.
- `source_text_1` … `source_text_4` — reference/source material the essay is based on (for source-dependent responses). These are useful later for a *relevance* feature (comparing the essay against its source), but are **not** used as direct ML input features in the Review 1 baseline.

**Columns explicitly excluded as predictive features**
- `economically_disadvantaged`, `student_disability_status`, `ell_status`, `race_ethnicity`, `gender`

**Why demographic columns are excluded:**
1. **Fairness/ethics** — using protected demographic attributes to predict a score that affects a student's grade risks encoding and amplifying bias (e.g., systematically scoring one group lower/higher regardless of essay quality).
2. **Construct validity** — an essay-scoring model should score the *essay*, not the *essay writer's demographics*. Including them would let the model "cheat" by learning demographic shortcuts instead of genuine writing-quality signal.
3. **Generalization** — a model that leans on demographic columns will not generalize to real deployment, where such data may be unavailable, inconsistent, or legally sensitive to collect/use.

These columns are kept in the raw dataframe only for optional **post-hoc fairness auditing** (e.g., checking whether prediction errors are evenly distributed across groups) — never as model input.

## 6. Problem type (Review 1 — Step 1.5)

The `score` column above is numeric (float) with a small number of ordered integer-like values (a bounded scale, min/max shown above). This is characteristic of an **ordinal / bounded numeric rating**, and the standard, correct ML formulation for essay scoring is:

### Primary task: REGRESSION
We predict a continuous/ordinal score value. Regression is standard for AES because:
- The score has a genuine order and meaningful distance (a 4 is better than a 2 by roughly "2 points"), which classification would discard.
- Regression metrics (MAE, RMSE) penalize being "off by a little" less than being "off by a lot" — appropriate for grading.

### Why a confusion matrix does NOT belong to the main task
A confusion matrix is a **classification** evaluation tool — it compares predicted class vs. actual class. Plain regression output (continuous score) has no discrete "classes" to build a confusion matrix from, so generating one directly from regression predictions would be meaningless / fabricated.

### Secondary classification experiment (for the confusion matrix requirement only)
To satisfy the teacher's checklist item (confusion matrix), we will — *separately, and clearly labeled* — bucket the score into ordinal categories (e.g., Low / Medium / High) using a justified split (based on the actual observed score range from Section 6 above, not invented cut points) and train a simple classifier on those buckets. This is a **secondary demonstration only**; the main scoring system remains the regression model.

We will finalize the exact bucket boundaries once we see the full score distribution in `02_eda.ipynb`, since the boundaries must be justified by the *actual* data range, not guessed here.
